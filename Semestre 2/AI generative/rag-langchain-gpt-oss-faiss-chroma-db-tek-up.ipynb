```python
import fitz  # PyMuPDF 📄  
# PyMuPDF (aussi appelé fitz) est une bibliothèque permettant de manipuler et d'extraire du texte et des images à partir de fichiers PDF.
```

```python
# Utilisé pour diviser un texte en morceaux tout en maintenant un bon équilibre entre taille et contexte.

from langchain_huggingface import HuggingFaceEmbeddings
# Permet de transformer du texte en vecteurs numériques en utilisant un modèle de Hugging Face.


# Créer un modèle d'embedding pour convertir du texte en vecteurs numériques
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
# Ce modèle est optimisé pour capturer la signification des phrases de manière efficace.

# Fonction pour extraire le texte d'un PDF
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)  # Ouvre le fichier PDF
    text = "\n".join([page.get_text("text") for page in doc])  # Extrait et concatène le texte de toutes les pages
    return text  # Renvoie le texte complet du document

# Charger le PDF et extraire le texte
pdf_path = "TEKUP_Documentation_Académique_2026.pdf"  # Chemin du fichier PDF
pdf_text = extract_text_from_pdf(pdf_path)  # Extraction du texte du PDF

# Découper le texte en morceaux (chunks) pour faciliter le traitement
text_splitter = RecursiveCharacterTextSplitter(chunk_size=600, chunk_overlap=100)  
# chunk_size=600 : Chaque segment fera environ 600 caractères.
# chunk_overlap=100 : Un chevauchement de 100 caractères assure une transition fluide entre les segments.

chunks = text_splitter.split_text(pdf_text)  # Découpe le texte en morceaux

# Convertir chaque morceau en un objet Document pour LangChain
documents = [Document(page_content=chunk) for chunk in chunks]  
# LangChain utilise ces objets pour associer du texte à des métadonnées.

# ChromaDB est une base de données vectorielle utilisée pour stocker et rechercher des embeddings.
# Créer un index vectoriel Chroma avec les morceaux de texte
vectorstore = Chroma.from_documents(documents, embedding_model, persist_directory="./chroma_db")  
# Stocke les embeddings des textes dans une base de données Chroma.
```

```python
# Effectuer une recherche dans la base de données vectorielle  
retrieved_docs = vectorstore.similarity_search(
    "parle moi sur le Règlement des absences a tek up", k=1
)

# Afficher les résultats de la recherche du texte similaire bien reffiner en terme de visualisation
for i, doc in enumerate(retrieved_docs):
    print(f"Document {i+1}:\n{doc.page_content}\n{'-'*50}\n")

# Explication :
# - similarity_search() permet de rechercher les documents les plus similaires à la requête donnée.
# - La requête est une question sur l'architecture technique du chatbot.
# - k=3 signifie que l'on récupère les trois documents les plus pertinents en fonction de la similarité des vecteurs.
```

```python

# Créer un modèle de prompt pour intégrer le contexte et la question
prompt = PromptTemplate(
    input_variables=["context", "question"],  # Définir les variables d'entrée : "context" et "question"
    template="""Tu es TekUp AI Assistant.
Tu es un assistant académique officiel de TEK-UP.

Tu dois répondre uniquement en utilisant le contexte fourni.
Si la réponse n'existe pas dans le contexte, dis clairement que l'information
n’est pas disponible dans la documentation.
Ne fais aucune supposition.
Contexte :
{context}

Question :
{question}

Réponds de manière claire, concise et professionnelle.
 """  
# Structure du modèle de prompt
)

# Explication :
# - input_variables=["context", "question"] définit les variables que le modèle de prompt utilisera.
# - template est un modèle de texte qui inclut des espaces réservés pour insérer le contexte et la question.
#   Il servira à formuler des requêtes structurées pour l'IA, par exemple :
#   "Voici des informations pertinentes :\n\n[contexte]\n\nQuestion : [question]\n\nRéponse :"
#   L'IA remplira le modèle avec les valeurs fournies pour "context" et "question".
```

```python

# Créer un modèle GPT  pour la génération de texte
llm = ChatGroq(api_key=os.environ["GROQ_API_KEY"], model_name="openai/gpt-oss-120b")

# Initialiser l'historique de la conversation comme une liste vide
conversation_history = []

# Fonction qui utilise ChromaDB pour la recherche et GPT pour générer une réponse
def rag_chatbot(question):
    # Recherche des documents pertinents dans la base de données Chroma
    retrieved_docs = vectorstore.similarity_search(question, k=3)  # Recherche des documents les plus similaires à la question
    context = "\n".join([doc.page_content for doc in retrieved_docs])  # Rassemble le contenu des documents pertinents pour former le contexte
    
    # Ajouter la question à l'historique de la conversation
    conversation_history.append(f"Question: {question}")
    conversation_history.append(f"Context: {context}")
    #print(context)
    
    # Créer un prompt qui inclut l'historique de la conversation et la nouvelle question
    conversation_context = "\n".join(conversation_history)  # Concatène tout l'historique
    
    # Utiliser le prompt et le modèle pour générer la réponse
    response = prompt | llm  # Applique le modèle de prompt (fusionne le prompt et le modèle de génération)
    result = response.invoke({"context": conversation_context, "question": question})  # Génère la réponse en passant le contexte et la question comme entrées
    
    # Ajouter la réponse générée à l'historique de la conversation
    conversation_history.append(f"Réponse: {result.content}")
    
    return result.content  # Retourne la réponse générée par l'IA
```

```python
'''
"Quels sont les cycles de formation proposés à TEK-UP ?"
"Quelle est la note minimale pour valider un module ?"
"Le stage est-il obligatoire ?"
'''

# Tester le chatbot avec la question "C'est quoi LangChain ?"
question = "Le stage est-il obligatoire ?"  
# La question posée au chatbot. Vous demandez une explication détaillée des tendances mode et une synthèse des informations.

print(rag_chatbot(question))  
# Appel de la fonction rag_chatbot qui va rechercher des documents pertinents dans la base de données Chroma 
# et utiliser le modèle Gemini Pro pour générer une réponse à la question posée.
```

```python
import gradio as gr

# Questions test que vous souhaitez poser au chatbot
sample_questions = [
    "Quels sont les cycles de formation proposés à TEK-UP ?",
    "Le stage est-il obligatoire ?",
    "c'est quoi un LLM ?",
]

# Interface Gradio pour le chatbot
interface = gr.Interface(
    fn=rag_chatbot,  # Fonction qui sera appelée pour générer des réponses
    inputs=gr.Textbox(label="Posez une question", lines=5),  # Entrée de la question
    outputs=gr.Textbox(label="Réponse", lines=5),  # Affiche la réponse générée dans une boîte de texte avec 5 lignes
    live=False,  # Met à jour la réponse en temps réel
    title="TekUp AI Assistant",  # Titre de l'interface
    description="Posez des questions sur TEK-UP et obtenez des réponses détaillées en fonction du contexte.",
    flagging_mode="never",  # Pour ne pas activer le marquage des entrées par défaut
    examples=sample_questions  # Liste des exemples de questions
)

# Lancer l'interface Gradio
interface.launch()
```
