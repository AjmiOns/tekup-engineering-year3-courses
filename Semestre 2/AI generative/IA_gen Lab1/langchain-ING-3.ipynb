```python
# Import prompt and define PromptTemplate

template = """
You are an expert data scientist with an expertise in building deep learning models. 
Explain the concept of {concept} in a couple of lines
"""

prompt = PromptTemplate(
    input_variables=["concept"],
    template=template,
)
```

```python
# Displaying results in markdown format library
from IPython.display import Markdown, display

# Run LLM with PromptTemplate

result = llm.invoke(prompt.format(concept="llm"))
display(Markdown(result.content))
```

```python
# define the chain between the language model (LLM) and prompt as arguments.
from IPython.display import Markdown, display


#LCLE langchain chain language expression
chain = prompt | llm

# Run the chain only specifying the input variable.
response = chain.invoke({"concept": "model d'embedding"})
display(Markdown(response.content))
```

```python
# Define a second prompt 

second_prompt = PromptTemplate(
    input_variables=["ml_concept"],
    template="Turn the concept description of {ml_concept} and explain it to me like I'm five in 500 words",
)
chain_two = second_prompt | llm
```

```python
# Define a sequential chain using the two chains above: the second chain takes the output of the first chain as input

from langchain_core.runnables import RunnablePassthrough

# Create a chain that passes the output of the first chain to the second chain
overall_chain = chain | (lambda x: {"ml_concept": x.content}) | chain_two

# Run the chain specifying only the input variable for the first chain.
explanation = overall_chain.invoke({"concept": "LLM"})
display(Markdown(explanation.content))
```

```python
# Import utility for splitting up texts and split up the explanation given above into document chunks

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap  = 0,
)

texts = text_splitter.create_documents([explanation.content])
print(f"Number of chunks: {len(texts)}")
for i, t in enumerate(texts):
    print(f"--- Chunk {i} ---")
    print(t.page_content,end='\n\n')
```

```python

from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
# Turn the first text chunk into a vector with the embedding

query_result = embeddings.embed_query(texts[0].page_content)
print(query_result)
print(f"Length of the embedding vector: {len(query_result)}")
```

```python
# Import and initialize ChromaDB

from langchain_community.vectorstores import Chroma

# ChromaDB will store data locally by default
# No API key needed for local usage
```

```python
# Create vector store with ChromaDB and add documents
search = Chroma.from_documents(
	texts,
	embeddings,
	persist_directory="./chroma_db"  # Optional: specify where to save the database locally
)
```

```python
# Do a simple vector similarity search + generate a response based on the retrieved content

# Define the query and perform similarity search
query = "What is magical about an autoencoder?"
result = search.similarity_search(query, k=2)

print(f"Retrieved :  {result} .")

# Define a prompt that takes both the query and retrieved content as input
template = """
you are the Generator into a rag Pipeline and you recieve this query : {query}  
and this is  the result of the retrieved content : {retrived_content}
you sole task is to generate directly the user is response :
"""

prompt = PromptTemplate(
    input_variables=["query","retrived_content" ],
    template=template,
)

# define the chain between the language model (LLM) and prompt as arguments.
chain = prompt | llm

#print(result)
# Do the inference with the chain
response = chain.invoke({"query": query, "retrived_content": result[0].page_content})
display(Markdown(response.content))
```

```python
# Execute the Python agent
# Note: Using a different model that supports function calling better
from langchain_groq import ChatGroq

# Create a new LLM instance with a model that supports tools
llm_for_agent = ChatGroq(api_key=os.environ["GROQ_API_KEY"], model_name="llama-3.3-70b-versatile")

# create the agent with the new LLM
# Create a calculator agent using Python REPL (REPL: Read-Eval-Print Loop, an interactive programming environment
# similar to a terminal or command line will be used by the agent to perform calculations)
from langchain_experimental.agents.agent_toolkits import create_python_agent
from langchain_experimental.tools import PythonREPLTool

agent_executor = create_python_agent(
	llm=llm_for_agent,
	tool=PythonREPLTool(),
	verbose=True
)

agent_executor.run("Find the roots (zeros) if the quadratic function 3 * x**2 + 2*x -1")
```

```python
# Since we already have llm_for_agent and the necessary imports, 
# we can directly use the Python agent which can handle an other  calculations

calculator_agent_executor = create_python_agent(
    llm=llm_for_agent,
    tool=PythonREPLTool(),
    verbose=True
)

# Run the agent with a calculation task
calculator_agent_executor.invoke({"input": "Calculate the area of a circle with radius 5 using the formula pi * r^2, where pi is approximately 3.14159"})
```
