```python
import os
import gradio as gr
import requests
import json
```

```python
# Function to call Groq API
def call_groq(prompt, api_key, model="llama-3.1-8b-instant", temperature=0.7):
    """Call the Groq API with the given prompt and return the response."""
    if not api_key:
        return "Please provide a Groq API key."
    
    headers = {
        "Authorization": f"Bearer {api_key}",
        "Content-Type": "application/json"
    }
    
    data = {
        "model": model,
        "messages": [{"role": "user", "content": prompt}],
        "temperature": temperature,
        "max_tokens": 1000
    }
    
    try:
        response = requests.post(
            "https://api.groq.com/openai/v1/chat/completions",
            headers=headers,
            data=json.dumps(data)
        )
        
        if response.status_code == 200:
            return response.json()["choices"][0]["message"]["content"]
        else:
            return f"Error: {response.status_code} - {response.text}"
    except Exception as e:
        return f"Error: {str(e)}"

# Available Groq models
groq_models = [
    "llama-3.1-8b-instant",
    "llama-3.3-70b-versatile",
    "gemma2-9b-it",
    "mixtral-8x7b-32768",
    "qwen-qwq-32b",
    "mistral-saba-24b"
]
```

```python
# Create Gradio interface
def groq_interface(prompt, api_key, model, temperature):
    return call_groq(prompt, api_key, model, temperature)

# Build the Gradio app
with gr.Blocks(title="Groq AI Interface at tek up") as app:
    gr.Markdown("# Simple Groq AI Interface")
    gr.Markdown("Enter your prompt and Groq API key to get a response from the selected model.")
    
    with gr.Row():
        with gr.Column():
            api_key = gr.Textbox(
                label="Groq API Key", 
                placeholder="Enter your Groq API key here...",
                type="password"
            )
            model = gr.Dropdown(
                choices=groq_models,
                label="Select Model",
                value="llama-3.1-8b-instant"
            )
            temperature = gr.Slider(
                minimum=0.0,
                maximum=1.0,
                value=0.7,
                step=0.1,
                label="Temperature",
                info="Higher values (0.8) = more creative, Lower values (0.2) = more focused"
            )
            prompt = gr.Textbox(
                label="Your Prompt", 
                placeholder="Enter your prompt here...",
                lines=5
            )
            submit_btn = gr.Button("Submit")
        
        with gr.Column():
            output = gr.Textbox(label="Response", lines=10)
    
    submit_btn.click(
        fn=groq_interface,
        inputs=[prompt, api_key, model, temperature],
        outputs=output
    )
    
    gr.Markdown("""
    ## Notes
    - You need a Groq API key to use this interface. Get one at [groq.com](https://console.groq.com/keys).
    - Temperature controls randomness: higher values (0.8+) give more creative responses, lower values (0.2-) give more focused responses.
    - The "llama-3.1-8b-instant" model is fast and efficient for most queries.
    - For more complex tasks, try the "llama-3.3-70b-versatile" model.
    """)

# Launch the app
if __name__ == "__main__":
    app.launch()
    
print("Gradio app is ready to launch. Run this script to start the interface.")
```
